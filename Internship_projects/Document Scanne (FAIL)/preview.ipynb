```python
import pyarrow.parquet as pq

# Get file metadata
parquet_file = pq.ParquetFile('0000.parquet')
print(parquet_file.schema)
print(f"Number of rows: {parquet_file.metadata.num_rows}")
print(f"Columns: {parquet_file.schema.names}")
```

```python
import pandas as pd

# Read only the ground_truth column (super fast, no images loaded)
df_labels = pd.read_parquet('0000.parquet', columns=['ground_truth'])
print(df_labels['ground_truth'].value_counts())

# Or get unique labels
unique_labels = df_labels['ground_truth'].unique()
print(f"Found {len(unique_labels)} unique labels:")
print(unique_labels)
```
