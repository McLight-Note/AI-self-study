```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Scatter Plot
x_data = np.random.random(1000) * 100
y_data = np.random.random(1000) * 100

plt.scatter(x_data, y_data, c='red', marker='*', s=150, alpha=0.3)
# s      - size
# alpha  - opacity if there are a lot of dots
# marker - how the dots should be
# c      - colors, you can use hex colors '#000, #00f'
```

```python
# Line chart
years = [2006 + x for x in range(20)]
weights = [80, 83, 84, 85, 86, 82, 81, 79, 80, 83, 82, 84, 85, 86, 88, 90, 95, 100, 102, 104]
plt.plot(years, weights, c='g', lw=3, linestyle='--')

# lw        - linewidth
# linestyle - how we define the line
```

```python
# Line chart
years = [2006 + x for x in range(20)]
weights = [80, 83, 84, 85, 86, 82, 81, 79, 80, 83, 82, 84, 85, 86, 88, 90, 95, 100, 102, 104]
plt.plot(years, weights, 'b--', lw=3)

# 'r--' - we can also use it like this for 'red' + '--'
# 'b--' - we can also use it like this for 'blue' + '--'
```

```python
# Bar chart

x = ['c++', 'c#', 'Python', 'Java', 'Go']
y = [20,50,140,1, 45]

plt.bar(x,y, color='r', align='edge', width=0.5, edgecolor='green', lw=6)

# align     - to define the lower text
# width     - to define the width of the bars
# edgecolor - is the color of the edge of the bars
# lw        - linewidth of the edges of the bars
```

```python
# Histogram

ages = np.random.normal(20, 1.5, 1000)
plt.hist(ages, bins=20, cumulative=True)

# np.random.normal                      - is a normal distribution for the random numbers in numpy
# bins                                  - how many boxes you want to see
# bins=[ages.min(), 18, 21, ages.max()] - you can pass a formula to get the specified bin
# cumulative=True                       - for acsending order
```

```python
langs = ['Python', 'C++', 'Java', 'Go', 'C#']
votes = [50, 24, 14, 6, 17]

explodes = [0,0,0,0.2,0]

plt.pie(votes, labels=langs, explode=explodes, autopct='%.2f%%', pctdistance=1.3, startangle=90)

# explode=explodes[n] - taking some parts of the pie chart out of it, n=[list]
# autopct=''          - to mean the percentile of the pie chart blocks
# pctdistance=1.3     - distance from the center of the chart
# startangle=90       - starting angle of the original pie chart
```

```python
heights = np.random.normal(172,8,300)
plt.boxplot(heights)
```

```python
first = np.linspace(0,10,25)
second = np.linspace(10,200,25)
third = np.linspace(200, 210, 25)
fourth = np.linspace(210, 230, 25)

data = np.concatenate((first, second, third, fourth))
plt.boxplot(data)
```

```python
years = [2014,2015,2016,2017,2018,2019,2020,2021]
income = [55,56,62,61,72,72,73,75]
income_ticks = list(range(50,81,2))

plt.title('Income of John (in USD)', fontsize=15, fontname="FreeSerif")
plt.xlabel('Year')
plt.ylabel('Yearly income in USD')
plt.yticks(income_ticks, [f"${x}k" for x in income_ticks])
plt.plot(years, income)
```
