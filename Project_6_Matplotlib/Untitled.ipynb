```python
 import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import pandas as pd

%reload_ext autoreload
%autoreload 2
```

```python
x_1 = np.linspace(0,5,10)
y_1 = x_1 ** 2
plt.plot(x_1,y_1)
plt.title('Days Squared Chart')
plt.xlabel('Days')
plt.ylabel('Days Squared')
```

```python
plt.subplot(1,2,1)
plt.plot(x_1, y_1,"r")
plt.title('Days Squared Chart')
plt.xlabel('Days')
plt.ylabel('Days Squared')
plt.subplot(1,2,2)
plt.plot(x_1, y_1,"b")
```

```python
figure_1 = plt.figure(figsize=(5,4), dpi=100)
axes_1 = figure_1.add_axes([0.1,0.1,0.9,0.9])
axes_1.set_xlabel('Days')
axes_1.set_ylabel('Days Squared')
axes_1.set_title('Days Squared Chart')
axes_1.plot(x_1,y_1,label='x/(x**2)')
axes_1.plot(y_1,x_1,label='(x**2)/x')
axes_1.legend(loc=0)

axes_2 = figure_1.add_axes([0.45, 0.45, 0.4, 0.3])
axes_2.set_xlabel('Days')
axes_2.set_ylabel('Days Squared')
axes_2.set_title('Days Squared Chart')
axes_2.plot(x_1,y_1, 'g')

axes_2.text(0,40,"Message")
```

```python
fig_2, axes_2 = plt.subplots(figsize=(8,4), nrows=1, ncols=4)
plt.tight_layout()
axes_2[0].set_title('Plot 1')
axes_2[0].set_xlabel('X')
axes_2[0].set_ylabel('X Squared')
axes_2[0].plot(x_1,y_1)
axes_2[0].plot(y_1,x_1)
axes_2[1].set_title('Plot 2')
axes_2[1].set_xlabel('X')
axes_2[1].set_ylabel('X Squared')
axes_2[1].plot(x_1,y_1, 'g')
axes_2[2].set_title('Plot 3')
axes_2[2].set_xlabel('X')
axes_2[2].set_ylabel('X Squared')
axes_2[2].plot(x_1,y_1, 'r')
axes_2[3].set_title('Plot 4')
axes_2[3].set_xlabel('X')
axes_2[3].set_ylabel('X Squared')
axes_2[3].plot(x_1,y_1, 'p')
```

```python
fig_3 = plt.figure(figsize=(6,4))
axes_3 = fig_3.add_axes([0,0,1,1])

# This is some styling options
axes_3.plot(x_1,y_1, color='navy', alpha=.75, lw=5, ls='-.', marker='o', markersize=10, markerfacecolor='y',
            markeredgecolor='r', markeredgewidth=3)

# This is to zoom in
axes_3.set_xlim(0,3)
axes_3.set_ylim(0,8)

# This is to put the backgound grid
axes_3.grid(True, color='0.6', dashes=(5,2,1,2))
axes_3.set_facecolor('#FAEBD7')
```

```python
fig_3.savefig('3rd_plot.png')
```
