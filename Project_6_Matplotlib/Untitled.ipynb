```python
import numpy as np
import matplotlib.pyplot as plt

x_1 = np.linspace(0, 10, 100)
y_1 = x_1 ** 2

fig_2, axes_2 = plt.subplots(figsize=(8, 4), nrows=1, ncols=4)
plt.tight_layout()

axes_2[0].set_title('Plot 1')
axes_2[0].set_xlabel('X')
axes_2[0].set_ylabel('X Squared')
axes_2[0].plot(x_1, y_1)
axes_2[0].plot(y_1, x_1)

axes_2[1].set_title('Plot 2')
axes_2[1].set_xlabel('X')
axes_2[1].set_ylabel('X Squared')
axes_2[1].plot(x_1, y_1, 'g')

axes_2[2].set_title('Plot 3')
axes_2[2].set_xlabel('X')
axes_2[2].set_ylabel('X Squared')
axes_2[2].plot(x_1, y_1, 'r')

axes_2[3].set_title('Plot 4')
axes_2[3].set_xlabel('X')
axes_2[3].set_ylabel('X Squared')
axes_2[3].plot(x_1, y_1, 'm')

arr_1 = np.random.randint(1, 6, 2000)
arr_2 = np.random.randint(1, 6, 2000)
arr_3 = arr_1 + arr_2

for ax in axes_2:
    ax.hist(arr_3, bins=7, density=False, stacked=True, histtype='step', color='orange')

plt.show()
```

```python
import random

fig_6 = plt.figure(figsize= (8,5) ,dpi=100)
axes_6 = fig_6. add_axes ([0.1,0.1,0.9,0.91])

# Create a pie chart of the number of Pokemon by type
types = [ 'Water', 'Ground', 'Normal', 'Flying', 'Grass', 'Psychic', 'Bug', 'Fire', 'Poison', 'Rock',
          'Ghost', 'Fighting', 'Dark', 'Steel', 'Electric', 'Dragon', 'Fairy', 'Ice']
poke_num = [133, 109, 101, 98, 85, 77, 68, 66, 65, 60, 57, 54, 53, 51, 50, 50, 46, 40]
colors = []
for i in range(18):
    rgb = random.uniform(0,.5), random.uniform(0,.5), random.uniform(0,.5)
    colors.append(rgb)

explode = [0] * 18
explode[0] = 0.2
wedges, texts, autotexts = plt.pie(poke_num, explode=explode, labels=types, colors=colors, autopct='%1.0f%%',
                                    shadow=True, startangle=140, textprops=dict(color='w'))
plt.legend(wedges, types, loc='right', bbox_to_anchor=(1, 0, 0.5, 1))
```
