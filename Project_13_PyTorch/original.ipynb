### Tensor datatypes
Aside from shape issues (tensor shapes don't match up), two of the other most common issues you'll come across in PyTorch are datatype and device issues.

```python
float_32_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=None, 
                                                device='cpu',
                                                requires_grad=False)
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_64_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=torch.float64, 
                                                device='cpu',
                                                requires_grad=False)
float_64_tensor
```

```python
float_64_tensor.dtype
```

```python
float_32_tensor * float_64_tensor
```

```python
float_16_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=torch.float16, 
                                                device='cpu',
                                                requires_grad=False)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
float_16_tensor * float_64_tensor
```

```python
int_32_tensor = torch.tensor([3,6,9], dtype=torch.int,)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

### Manipulation tensors

1. The inner dimensions must match
2. The shape of matrix multiplication is the **outer dimension**

```python
# Addition

t1 = torch.tensor([1,2,3])
t1 + 100
```

```python
# Multiplication
t1 * 12
```

```python
t1 - 10
```

```python
t1 / 3
```

```python
print(torch.add(t1, 3))
```

```python
print(torch.mul(t1, 3))
```

```python
# Matrix multiplication (Elementwise)
t1 * t1
```

```python
%%time
# Matrix multiplication
torch.matmul(t1, t1)
```

```python
%%time
value = 0
for i in range(len(t1)):
    value == t1[i] ** t1[i]
print(value)
```

One of the most common errors in deep learning: shape errors

```python
# Shapes for matrix multiplications
t1 = torch.tensor([[1,2],[3,4],[5,6]])
t2 = torch.tensor([[7,8,9],[10,11,12]])
torch.mm(t1,t2) # torch.mm = torch.matmul
```

```python
# Transpose
t1 = torch.tensor([[1,2],[3,4],[5,6]])
t2 = torch.tensor([[7,8],[9,10],[11,12]])

print(t1.T)
print('\n', torch.mm(t1.T, t2), '\n')
print(torch.mm(t1, t2.T))
```
