## Creating a range of tensors and tensors-like

```python
# use torch.arange()
one_to_ten = torch.arange(start=0, end=1000, step=77)
one_to_ten
```

```python
# Tensor-like
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```

```python
ten_ones = torch.ones_like(input=one_to_ten)
ten_ones
```

### Tensor datatypes
Aside from shape issues (tensor shapes don't match up), two of the other most common issues you'll come across in PyTorch are datatype and device issues.

```python
float_32_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=None, 
                                                device='cpu',
                                                requires_grad=False)
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_64_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=torch.float64, 
                                                device='cpu',
                                                requires_grad=False)
float_64_tensor
```

```python
float_64_tensor.dtype
```

```python
float_32_tensor * float_64_tensor
```

```python
float_16_tensor = torch.tensor([3.0, 4.0, 9.0], dtype=torch.float16, 
                                                device='cpu',
                                                requires_grad=False)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
float_16_tensor * float_64_tensor
```

```python
int_32_tensor = torch.tensor([3,6,9], dtype=torch.int,)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```
