```python
import pandas as pd
```

```python
df = pd.read_csv('dirty_data.csv')
df
```

```python
df['Rank'] = df.index + 1
df = df.drop('Peak', axis=1)
df
```

```python
df = df.drop('All Time Peak', axis=1)
```

```python
df.columns = df.columns.str.strip()  # remove leading/trailing spaces
df.columns = df.columns.str.strip().str.replace('\xa0','')
df
```

```python
df['Actualgross'] = df['Actualgross'].str.replace('[^0-9$]', '', regex=True)
```

```python
df['Actualgross'] = df['Actualgross'].str.replace('$','', regex=False).str.replace(',','', regex=False)
df
```

```python
df['Actualgross'] = df['Actualgross'].apply(lambda x: f"{int(x):,}")
df
```

```python
df['Adjustedgross (in 2022 dollars)'] = df['Adjustedgross (in 2022 dollars)'].str.replace('$','', regex=False).str.replace(',','', regex=False)
df['Adjustedgross (in 2022 dollars)'] = df['Adjustedgross (in 2022 dollars)'].apply(lambda x: f"{int(x):,}")
df
```

```python
df['Tour title'] = df['Tour title'].str.replace(r'[\u2020\[\]].*$', '', regex=True).str.strip()
df
```

```python
df['Tour title'] = df['Tour title'].str.split('*').str[0].str.strip()
df
```

```python
df = df.drop('Ref.', axis=1)
df
```

```python
df['Average gross'] = df['Average gross'].str.replace('[^0-9$]', '', regex=True)
df['Average gross'] = df['Average gross'].str.replace('$','', regex=False).str.replace(',','', regex=False)
df['Average gross'] = df['Average gross'].apply(lambda x: f"{int(x):,}")
df
```

```python
df.columns = ['Rank', 'Actual Gross', 'Adjusted Gross', 'Artist', 'Tour Title', 'Year(s)', 'Shows', 'AverageGross']
df
```

```python
df.to_csv('cleaned_tours.csv', index=False)
```
