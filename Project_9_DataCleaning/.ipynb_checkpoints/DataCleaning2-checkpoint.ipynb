```python
import pandas as pd
```

```python
df = pd.read_csv('NFL.csv')
df = df.drop_duplicates()
```

```python
df = df.iloc[:, 0:9]
```

```python
df
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df_filtered = df[(df['Date'].dt.year >= 2010) & (df['Date'].dt.year <= 2015)]
df_filtered = df_filtered.reset_index(drop=True)
df_filtered
```

```python
df_filtered.fillna("")
```

```python
for x in df_filtered.index:
    if df_filtered.loc[x, 'TimeSecs'] <= 0:
        df_filtered.drop(x, inplace=True)
df_filtered
```

```python
df_filtered = df_filtered.reset_index(drop=True)
df_filtered
```

```python
df_filtered['down'] = df_filtered['down'].fillna("")
```
